# Visualize federated catalog results

Like the federated statistics visualization notebook, this notebook takes saved JSON and calls reusable display functions. It runs offline: no login, startup kit or running federation is needed. Only published schema metadata is visualized.

Run `prepare_data.py` and `job.py` first to produce `catalog_results.json`, or point the next cell at a saved result. The example has Parquet and CSV datasets at site-1, SQLite at site-2, and DuckDB at site-3.


```python
from IPython.display import display

from nvflare.app_opt.fdn.catalog.visualization import CatalogVisualization
```

## Load the catalog result

Set `CATALOG_FILE` to the `catalog_results.json` of a catalog read. When [get_catalog.ipynb](get_catalog.ipynb) runs this notebook, it supplies the fresh result path. Opened independently in a fresh kernel, this notebook uses the default below.


```python
CATALOG_FILE = globals().get("CATALOG_FILE", "/tmp/nvflare/federated-catalog/catalog_results.json")

vis = CatalogVisualization(CATALOG_FILE)
print(f"Catalog of Study {vis.study}")
```

## Sites and datasets

First each site's status in this read, then one row per site and dataset with its source, status and published table/column counts. A site can publish several datasets backed by different databases or file formats.


```python
display(vis.show_sites())
display(vis.show_datasets())
```

## Published tables

List the tables and their declared keys. Use `site` and `dataset` to select a source.


```python
display(vis.show_tables(dataset="omop"))
```

## Table schema

Inspect the columns, types, nullability and primary key of a table at one site.


```python
display(vis.show_schema("person", site="site-1", dataset="omop"))
```

## Compare tables across sites

Each cell shows the number of published columns. Red means the table was not published by that source; gray means its dataset was not read (unavailable or skipped). An unpublished table or column may have been excluded by the site's catalog rules.


```python
display(vis.show_table_comparison(dataset="omop"))
```

## Compare columns

See shared columns and columns published only by some sources. Yellow marks different reported values. Type names are shown as reported by each source.


```python
display(vis.show_column_comparison("condition_occurrence", dataset="omop"))
```

### A table present at each site


```python
display(vis.show_column_comparison("person", dataset="omop"))
```

### Compare nullability or primary keys

Use `field="nullable"` or `field="primary_key"`. Unknown nullability is distinct from a column that was not published.


```python
display(vis.show_column_comparison("person", dataset="omop", field="nullable"))
```

## Relationships

Show the foreign keys each source reports. Parquet files carry none; SQLite and DuckDB report the constraints declared in their tables.


```python
display(vis.show_relationships(dataset="omop"))
```

## Compare different datasets and databases

Leave `dataset` unset to compare all collections. The two column headers identify both site and dataset, so several databases at the same site remain distinct. Every `show_*` function also has a `get_*` counterpart that returns a DataFrame for further analysis.


```python
display(vis.show_table_comparison())
```
